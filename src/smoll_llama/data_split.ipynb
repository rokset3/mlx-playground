```python
with open("data/input.txt", "r") as f:
    input_text = f.readlines()
```

```python
text = [text for text in input_text if text != "\n"]
```

```python
len(text)
```

```python
train_size = 0.9
val_size = 0.05
test_size = 0.05
```

```python
total_size = len(text)
total_size * train_size
```

```python
train_ids_end = int(total_size * train_size)
val_ids_end = train_ids_end + int(total_size * val_size)
test_ids_end = total_size - 1
```

```python
with open("data/train_texts.txt", "w") as f:
    f.writelines(text[: train_ids_end])

with open("data/val_texts.txt", "w") as f:
    f.writelines(text[train_ids_end: val_ids_end])

with open("data/test_texts.txt", "w") as f:
    f.writelines(text[val_ids_end: test_ids_end])
```
